# SGMA GSA Metadata Builder

Builds a single canonical metadata table for every California Groundwater Sustainability Agency (GSA):

| source | fields |
|---|---|
| **DWR i03 Open Data CSV** (canonical listing) | `GSA_ID`, `GSA_Name`, `Basin_Subbasin_Name`, `POC_Name`, `POC_Email`, `POC_Phone`, `Posted_DT`, `Exclusive`, `Overlap`, `GSA_URL`, … |
| **SGMA Portal print page** (per-GSA scrape) | `formation_type` (SINGLE / MULTIPLE), `member_entities` (pipe-separated agencies) |

### Why this replaces the old listing scrape
The portal's old `/gsa/exportcsv` endpoint returns 404, and the DataTables-based listing pagination is fragile to scrape. The DWR i03 dataset on CNRA Open Data ships the full canonical listing — including the same numeric `GSA_ID` used in portal URLs — in a stable CSV download. We only hit the portal for the two Section C fields i03 doesn't carry.

### Output
`{repo_root}/data/core_data/metadata/sgma_gsa_full.csv` — one row per GSA. Resumable JSONL cache sits alongside it.

### Requirements
```
pip install requests pandas beautifulsoup4 tqdm
```


````python
import os, re, json, time, io
from pathlib import Path
import requests
import pandas as pd
from bs4 import BeautifulSoup
from tqdm.notebook import tqdm

# ─── Paths ─────────────────────────────────────────────────────────────
# Resolve repo root by walking up from the notebook until we hit a marker.
def _repo_root() -> Path:
    p = Path.cwd().resolve()
    for cand in [p, *p.parents]:
        if (cand / 'kings.Rproj').exists() or (cand / '.git').exists() and (cand / 'core_code').exists():
            return cand
    return p  # fallback

REPO_ROOT = _repo_root()
OUT_DIR   = REPO_ROOT / 'data' / 'core_data' / 'metadata'
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_CSV   = OUT_DIR / 'sgma_gsa_full.csv'
CACHE_JL  = OUT_DIR / 'gsa_section_c_cache.jsonl'

# ─── Endpoints ──────────────────────────────────────────────────────────
I03_URL   = ('https://gis.data.cnra.ca.gov/api/download/v1/items/'
             '51688234e6104427894328acee4983e8/csv?layers=0')
PORTAL    = 'https://sgma.water.ca.gov/portal'
HEADERS   = {'User-Agent': 'Mozilla/5.0 (compatible; UCD-research-scraper)',
             'Referer': f'{PORTAL}/gsa/all'}
DELAY_SEC = 0.4  # politeness between portal requests

print(f'REPO_ROOT : {REPO_ROOT}')
print(f'OUT_DIR   : {OUT_DIR}')
print(f'OUT_CSV   : {OUT_CSV}')
print(f'CACHE_JL  : {CACHE_JL}')
````

---
## Step 1 — Pull canonical GSA listing from DWR i03

i03 has one row per `(GSA, subbasin)` pair, so multi-area GSAs appear multiple times. We collapse to one row per `GSA_ID`, concatenating the subbasin names.


````python
resp = requests.get(I03_URL, headers=HEADERS, timeout=60)
resp.raise_for_status()
i03 = pd.read_csv(io.StringIO(resp.content.decode('utf-8-sig')))
print(f'i03 raw rows: {len(i03):,}  cols: {len(i03.columns)}')
print('columns:', list(i03.columns))

# Collapse to one row per GSA_ID. For columns that vary across rows of the same
# GSA (subbasin name, OBJECTID, areas), aggregate into a sorted pipe-separated list.
def _agg_unique(s):
    vals = sorted({str(v).strip() for v in s.dropna() if str(v).strip()})
    return ' | '.join(vals) if len(vals) > 1 else (vals[0] if vals else '')

per_gsa_cols = ['GSA_Name', 'Loc_GSA_ID', 'Basin_Name', 'Posted_DT', 'GSA_URL',
                'POC_Name', 'POC_Phone', 'POC_Email', 'Date_90_Days',
                'Exclusive', 'Overlap', 'COMMENTS', 'UPDATED_DT', 'MODIFIED_BY']
basin_cols   = ['Basin_Number', 'Basin_Subbasin_Number', 'Basin_Subbasin_Name']

agg = {c: 'first' for c in per_gsa_cols if c in i03.columns}
agg.update({c: _agg_unique for c in basin_cols if c in i03.columns})
agg['_n_basin_rows'] = ('OBJECTID', 'count') if False else 'size'  # placeholder; fixed below

df_main = (i03
           .groupby('GSA_ID', as_index=False)
           .agg({**{c: 'first' for c in per_gsa_cols if c in i03.columns},
                 **{c: _agg_unique for c in basin_cols if c in i03.columns}}))
df_main['n_basin_rows'] = i03.groupby('GSA_ID').size().reindex(df_main['GSA_ID']).values
df_main = df_main.sort_values('GSA_ID').reset_index(drop=True)

print(f'\nUnique GSAs after collapse: {len(df_main):,}')
df_main.head(3)
````

---
## Step 2 — Section C parser

Each `/portal/gsa/print/{id}` page has a section labelled **"CType of GSA Formation and Contact Information"** containing the contact name, a SINGLE/MULTIPLE marker, and (for MULTIPLE) the member-agency list. The page concatenates section letter + title with no space, e.g. `CType…`, `DMap…`, `ERequired…`.

The parser anchors on the section title text, then walks forward through the rendered text until the next section header (`^[A-Z][A-Z][a-z]…`). It returns the formation_type, member_entities, and a parser-side contact_name (we mainly rely on i03's `POC_Name`, but keep this for cross-check).


````python
SECTION_C_HDR = re.compile(r'C\s*Type of GSA Formation and Contact Information', re.I)
NEXT_SECTION  = re.compile(r'^[A-Z][A-Z][a-z]')  # 'DMap', 'ERequired', ...
FILE_LINE     = re.compile(r'\.(pdf|zip|docx?|xlsx?)\s*\(', re.I)
DATE_LINE     = re.compile(r'^\d{1,2}/\d{1,2}/\d{4}')
LIST_MARKER   = re.compile(r'^\s*(?:\d+[\.\)]|[\u2022\-\*])\s+')  # strip '1. ', '2) ', '- ', '* '

def _clean_lines(html_text):
    soup = BeautifulSoup(html_text, 'html.parser')
    out = []
    for raw in soup.get_text(separator='\n').splitlines():
        l = LIST_MARKER.sub('', raw).strip()
        if l:
            out.append(l)
    return out

def parse_section_c(html_text, gsa_name=None):
    """Extract formation_type, member_entities, contact_name from a GSA print page.
    Returns dict with keys formation_type, member_entities, contact_name, section_found."""
    out = {'formation_type': None, 'member_entities': None,
           'contact_name': None, 'section_found': False}
    lines = _clean_lines(html_text)

    start = next((i for i, l in enumerate(lines) if SECTION_C_HDR.search(l)), None)
    if start is None:
        return out
    out['section_found'] = True

    end = next((i for i, l in enumerate(lines[start+1:], start=start+1)
                if NEXT_SECTION.match(l) and not SECTION_C_HDR.search(l)),
               len(lines))
    block = [l for l in lines[start+1:end]
             if l != '-' and not FILE_LINE.search(l) and not DATE_LINE.match(l)]

    ft_idx = next((i for i, l in enumerate(block) if l.upper() in ('SINGLE', 'MULTIPLE')), None)
    if ft_idx is None:
        return out
    out['formation_type'] = block[ft_idx].upper()

    gsa_norm = (gsa_name or '').strip().lower()

    # Contact name: walk backwards from the SINGLE/MULTIPLE marker, find a
    # person-name-shaped line.
    for cand in reversed(block[:ft_idx]):
        if cand.lower() == gsa_norm:
            continue
        if cand.isupper() or cand.islower():
            continue
        words = cand.split()
        if not (2 <= len(words) <= 5):
            continue
        if any(ch.isdigit() for ch in cand):
            continue
        out['contact_name'] = cand
        break

    if out['formation_type'] == 'MULTIPLE':
        members = [l for l in block[ft_idx+1:]
                   if l.lower() != gsa_norm and l not in ('Yes', 'No', '-')]
        if members:
            out['member_entities'] = ' | '.join(members)
    return out
````

````python
# Unit tests: 244 = White Wolf (MULTIPLE), 481 = Arvin (SINGLE)
test_cases = {
    244: dict(name='White Wolf GSA', expect_type='MULTIPLE',
              expect_contact='Angelica Martin',
              expect_member_contains=['Arvin-Edison Water Storage District', 'Kern County']),
    481: dict(name='Arvin GSA', expect_type='SINGLE',
              expect_contact='Jeevan Muhar', expect_member_contains=None),
}

for gid, spec in test_cases.items():
    r = requests.get(f'{PORTAL}/gsa/print/{gid}', headers=HEADERS, timeout=20)
    parsed = parse_section_c(r.text, gsa_name=spec['name'])
    ok_type    = parsed['formation_type'] == spec['expect_type']
    ok_contact = (parsed['contact_name'] or '').strip() == spec['expect_contact']
    if spec['expect_member_contains']:
        members = parsed['member_entities'] or ''
        ok_members = all(m in members for m in spec['expect_member_contains'])
    else:
        ok_members = parsed['member_entities'] is None
    print(f'GSA {gid} ({spec["name"]}):')
    print(f'   type    : {parsed["formation_type"]!r}     {"OK" if ok_type else "FAIL"}')
    print(f'   contact : {parsed["contact_name"]!r}   {"OK" if ok_contact else "FAIL"}')
    print(f'   members : {(parsed["member_entities"] or "")[:80]!r}   {"OK" if ok_members else "FAIL"}')
    assert ok_type and ok_contact and ok_members, f'Parser failed on GSA {gid}'
    time.sleep(DELAY_SEC)
print('\nAll parser tests passed.')
````

---
## Step 3 — Scrape Section C for every GSA

Resumable: each result is appended to `gsa_section_c_cache.jsonl` so re-running picks up where it left off. Set `RESCRAPE_ALL = True` to ignore the cache.


````python
RESCRAPE_ALL = False

# Build name lookup so the parser can drop the GSA's own name from member lists
name_by_id = dict(zip(df_main['GSA_ID'].astype(int), df_main['GSA_Name']))

# Load existing cache
scraped = {}
if CACHE_JL.exists() and not RESCRAPE_ALL:
    with CACHE_JL.open() as f:
        for line in f:
            try:
                rec = json.loads(line)
                scraped[int(rec['gsa_id'])] = rec
            except Exception:
                continue
    print(f'Resuming — {len(scraped):,} GSAs already cached')

gsa_ids = sorted(name_by_id.keys())
todo    = [g for g in gsa_ids if g not in scraped]
print(f'Total GSAs: {len(gsa_ids):,}  |  to scrape: {len(todo):,}')

errors = []
mode   = 'w' if RESCRAPE_ALL else 'a'
with CACHE_JL.open(mode) as cache_f:
    for gid in tqdm(todo, desc='Section C scrape'):
        try:
            r = requests.get(f'{PORTAL}/gsa/print/{gid}', headers=HEADERS, timeout=20)
            r.raise_for_status()
            parsed = parse_section_c(r.text, gsa_name=name_by_id.get(gid))
            rec = {'gsa_id': int(gid), **parsed}
            scraped[gid] = rec
            cache_f.write(json.dumps(rec) + '\n')
            cache_f.flush()
            time.sleep(DELAY_SEC)
        except Exception as e:
            errors.append({'gsa_id': gid, 'error': str(e)})

print(f'\nDone. errors: {len(errors)}')
if errors:
    print('Failed IDs (first 20):', [e["gsa_id"] for e in errors[:20]])
````

---
## Step 4 — Merge & save


````python
df_section_c = pd.DataFrame(list(scraped.values()))
if not df_section_c.empty:
    df_section_c['gsa_id'] = df_section_c['gsa_id'].astype(int)
df_main['GSA_ID'] = df_main['GSA_ID'].astype(int)

df_full = df_main.merge(
    df_section_c.rename(columns={'gsa_id': 'GSA_ID',
                                 'contact_name': 'section_c_contact_name'}),
    on='GSA_ID', how='left')

df_full.to_csv(OUT_CSV, index=False)
print(f'Saved {len(df_full):,} rows → {OUT_CSV}')
print('\nFormation type breakdown:')
print(df_full['formation_type'].value_counts(dropna=False))

preview_cols = ['GSA_ID', 'GSA_Name', 'Basin_Name', 'POC_Name',
                'formation_type', 'member_entities']
df_full[preview_cols].head(8)
````

---
## Step 5 — Verification

Sanity checks on the merged output. These should all pass; assertion failures highlight scrape or merge problems.


````python
import random

problems = []

# 1. Row count parity
if len(df_full) != len(df_main):
    problems.append(f'row count mismatch: df_full={len(df_full)} vs df_main={len(df_main)}')

# 2. Value domain on formation_type
bad_ft = set(df_full['formation_type'].dropna().unique()) - {'SINGLE', 'MULTIPLE'}
if bad_ft:
    problems.append(f'unexpected formation_type values: {bad_ft}')

# 3. MULTIPLE GSAs should have member_entities populated
missing_members = df_full[(df_full['formation_type'] == 'MULTIPLE') &
                          (df_full['member_entities'].isna() |
                           (df_full['member_entities'] == ''))]
if len(missing_members) > 0:
    problems.append(f'{len(missing_members)} MULTIPLE GSAs have empty member_entities')
    print('  first 5 IDs:', missing_members['GSA_ID'].head().tolist())

# 4. SINGLE GSAs should NOT have member_entities
extra_members = df_full[(df_full['formation_type'] == 'SINGLE') &
                        df_full['member_entities'].notna() &
                        (df_full['member_entities'] != '')]
if len(extra_members) > 0:
    problems.append(f'{len(extra_members)} SINGLE GSAs unexpectedly have member_entities')

# 5. Cross-check contact name: when the page parser found a name, it should
#    usually match the i03 POC_Name. We just report disagreement rate.
both = df_full.dropna(subset=['POC_Name', 'section_c_contact_name'])
if len(both):
    match = (both['POC_Name'].str.strip().str.lower()
             == both['section_c_contact_name'].str.strip().str.lower())
    print(f'contact-name agreement: {match.sum()}/{len(both)} '
          f'({100*match.mean():.1f}%) — disagreements are OK if i03 has been updated')

# 6. Spot-check: re-fetch 3 random IDs and confirm the live page agrees with cache
random.seed(0)
sample_ids = random.sample(list(scraped.keys()), k=min(3, len(scraped)))
for gid in sample_ids:
    r = requests.get(f'{PORTAL}/gsa/print/{gid}', headers=HEADERS, timeout=20)
    live = parse_section_c(r.text, gsa_name=name_by_id.get(gid))
    cached = scraped[gid]
    ok = (live['formation_type'] == cached.get('formation_type')
          and live['member_entities'] == cached.get('member_entities'))
    print(f'  spot-check GSA {gid}: {"OK" if ok else "DIVERGED"}')
    time.sleep(DELAY_SEC)

print('\n--- summary ---')
if problems:
    for p in problems:
        print(' ⚠', p)
else:
    print(' ✓ All checks passed.')
````
